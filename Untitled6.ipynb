```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import matplotlib.pyplot as plt
```

```python
reviews = {
    "text": [
        "The movie was wonderful and inspiring",
        "I really enjoyed this film a lot",
        "Such a beautiful and touching story",
        "Amazing direction and great acting",
        "I loved every minute of this movie",
        "What a fantastic experience overall",

        "The movie was boring and too long",
        "I really disliked this film so much",
        "Such a disappointing and dull story",
        "Terrible acting and bad direction",
        "I hated every minute of this movie",
        "What an awful experience overall"
    ],
    "sentiment": [
        "positive","positive","positive","positive","positive","positive",
        "negative","negative","negative","negative","negative","negative"
    ]
}
```

```python
df = pd.DataFrame(reviews)
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df["text"])
y = df["sentiment"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=10, stratify=y
)
```

```python
from sklearn.metrics import classification_report
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix

mnb = MultinomialNB()
mnb.fit(X_train, y_train)

y_pred = mnb.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
vectorizer_bin = CountVectorizer(binary=True)
X_bin = vectorizer_bin.fit_transform(df["text"])
```

```python
Xb_train, Xb_test, yb_train, yb_test = train_test_split(
    X_bin, y, test_size=0.3, random_state=42
)
```

```python
from sklearn.naive_bayes import BernoulliNB
bnb = BernoulliNB()
bnb.fit(Xb_train, yb_train)
bnb_pred = bnb.predict(Xb_test)

print("\n=== BERNOULLI NAIVE BAYES ===")
print("Accuracy:", accuracy_score(yb_test, bnb_pred))
print("Confusion Matrix:")
print(confusion_matrix(yb_test, bnb_pred))
```

```python
feature_names = vectorizer.get_feature_names_out()
```

```python
pos_text = df[df.sentiment=="positive"]["text"]
neg_text = df[df.sentiment=="negative"]["text"]
```

```python
pos_count = vectorizer.transform(pos_text).sum(axis=0).A1
neg_count = vectorizer.transform(neg_text).sum(axis=0).A1
```

```python

N = 8
```

```python
top_pos = np.argsort(pos_count)[-N:]
top_neg = np.argsort(neg_count)[-N:]
```

```python
plt.figure(figsize=(10,4))
plt.bar([feature_names[i] for i in top_pos], pos_count[top_pos])
plt.xticks(rotation=45)
plt.title("Positive Words")
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.bar([feature_names[i] for i in top_neg], neg_count[top_neg])
plt.xticks(rotation=45)
plt.title("Negative Words")
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
```

```python
comments = {
    "text": [
        "Service was excellent and friendly",
        "I am very satisfied with this",
        "Really good quality and fast delivery",
        "Super happy with the purchase",
        "I would definitely recommend this",
        "Everything about this was great",

        "Service was rude and slow",
        "I am very unhappy with this",
        "Really poor quality and late delivery",
        "Not happy with the purchase at all",
        "I would never recommend this",
        "Everything about this was bad"
    ],
    "label": [
        "positive","positive","positive","positive","positive","positive",
        "negative","negative","negative","negative","negative","negative"
    ]
}
```

```python
comments = {
    "text": [
        "Service was excellent and friendly",
        "I am very satisfied with this",
        "Really good quality and fast delivery",
        "Super happy with the purchase",
        "I would definitely recommend this",
        "Everything about this was great",

        "Service was rude and slow",
        "I am very unhappy with this",
        "Really poor quality and late delivery",
        "Not happy with the purchase at all",
        "I would never recommend this",
        "Everything about this was bad"
    ],
    "label": [
        "positive","positive","positive","positive","positive","positive",
        "negative","negative","negative","negative","negative","negative"
    ]
}
```

```python
df2 = pd.DataFrame(comments)
```

```python
bin_vec = CountVectorizer(binary=True)
Xb = bin_vec.fit_transform(df2["text"])
yb = df2["label"]
```

```python
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    Xb, yb, test_size=0.25, random_state=5, stratify=yb
)
```

```python
bnb = BernoulliNB()
bnb.fit(X_train_b, y_train_b)
```

```python
y_pred_b = bnb.predict(X_test_b)
```

```python
print("Accuracy:", accuracy_score(y_test_b, y_pred_b))
print(confusion_matrix(y_test_b, y_pred_b))
```

```python
feature_names_b = bin_vec.get_feature_names_out()
log_prob = bnb.feature_log_prob_
```

```python
classes = bnb.classes_
idx_neg = list(classes).index("negative")
idx_pos = list(classes).index("positive")
```

```python
prob_neg = np.exp(log_prob[idx_neg])
prob_pos = np.exp(log_prob[idx_pos])
```

```python

N=8
```

```python
top_pos = np.argsort(prob_pos)[-N:]
top_neg = np.argsort(prob_neg)[-N:]
```

```python
plt.figure(figsize=(10,4))
plt.bar([feature_names_b[i] for i in top_pos], prob_pos[top_pos])
plt.xticks(rotation=45)
plt.title("Positive Words (Bernoulli)")
plt.show()
```

```python
plt.figure(figsize=(10,4))
plt.bar([feature_names_b[i] for i in top_neg], prob_neg[top_neg])
plt.xticks(rotation=45)
plt.title("Negative Words (Bernoulli)")
plt.show()
```
